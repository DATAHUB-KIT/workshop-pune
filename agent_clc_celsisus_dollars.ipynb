```python
from google.colab import userdata
```

```python
GEMINI_API_KEY=userdata.get('gemini9')
```

```python
client=genai.Client(api_key=GEMINI_API_KEY)
```

```python
def calculator(expression: str) -> str:
    return str(eval(expression))
```

```python
response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents="What is 25*10?",
    config={
        "tools": [calculator]
    }
)
```

```python
print(response.text)
```

```python
def add_numbers(a: float, b: float) -> float:
        return a + b
```

```python
def celsius_to_fahrenheit(celsius: float) -> float:
        return (celsius * 9/5) + 32
```

```python
def dollars_to_rupees(dollars: float) -> float:

    exchange_rate = 85
    return dollars * exchange_rate
```

```python
from google.genai import types

config = types.GenerateContentConfig(
    system_instruction="""
    You are a beginner-friendly AI assistant.

    Use the correct tool for the user question.


    Explain the answer in simple words.
    """,

    tools=[
        add_numbers,
        celsius_to_fahrenheit,
        dollars_to_rupees
    ]
)
```

```python
while True:

    question = input("\nYou: ")

    if question.lower() == "exit":
        print("Agent: Goodbye!")
        break

    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=question,
        config=config
    )

    print("Agent:", response.text)
```
